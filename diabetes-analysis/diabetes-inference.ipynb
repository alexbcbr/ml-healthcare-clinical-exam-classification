# Testing Diabetes deployed model

```python
#ENDPOINT = "diabetes-readmission-api"
ENDPOINT = "diabetes-readmission"

sess = sagemaker.Session()

predictor = RealTimePredictor(endpoint=ENDPOINT, 
                              sagemaker_session=sess,
                              content_type="text/csv",
                              accept="text/csv")
```

```python
def predict_diabetes(datapatiente):
    response = predictor.predict(datapatiente)
    result = json.loads(response)
    #print('Probabilities for all classes: ', response)
    #print(result)
    if result == 0:
        print('NO readmission predicted')
    else:
        print('Readimission')
```
