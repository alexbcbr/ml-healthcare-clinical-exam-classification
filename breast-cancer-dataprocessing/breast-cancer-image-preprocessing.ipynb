# Reading files directly from S3

```python
import boto3
from io import BytesIO
import matplotlib.image as mpimg
import matplotlib.pyplot as plt
```

```python
resource = boto3.resource('s3', region_name='us-east-2')
bucket = resource.Bucket('alexbc.datalake.us-east-2')
```

```python
image_object = bucket.Object('breast-cancer-histopathology/curated-data/0/10253_idx5_x1001_y1001_class0.png')
image = mpimg.imread(BytesIO(image_object.get()['Body'].read()), 'png')

plt.figure(0)
plt.imshow(image)
```

```python
image_object = bucket.Object('breast-cancer-histopathology/curated-data/1/10253_idx5_x501_y351_class1.png')
image = mpimg.imread(BytesIO(image_object.get()['Body'].read()), 'png')

plt.figure(0)
plt.imshow(image)
```
