# Covert image into RecordID format

## Get the im2rec script from Apache MXNet Github repository

```python
%%bash

wget https://raw.githubusercontent.com/apache/incubator-mxnet/master/tools/im2rec.py
chmod +x im2rec.py
```

```python
%%bash

python im2rec.py --list --recursive --test-ratio 0.3 --train-ratio 0.7 data data/
```

```python
%%bash

python im2rec.py --num-thread 4 --pass-through data_train.lst data/
python im2rec.py --num-thread 4 --pass-through data_test.lst data/
```

```python
%%bash

aws s3 cp data_train.rec s3://alexbc.datalake.us-east-2/breast-cancer-histopathology/curated-data/recordio/
aws s3 cp data_test.rec s3://alexbc.datalake.us-east-2/breast-cancer-histopathology/curated-data/recordio/
```
